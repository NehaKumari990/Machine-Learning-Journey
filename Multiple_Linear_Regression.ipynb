```python

import numpy as np

# Training data
X = np.array([
    [2, 60, 50],
    [3, 70, 60],
    [4, 80, 70],
    [5, 90, 80],
    [6, 95, 90]
], dtype=float)

y = np.array([45, 55, 65, 75, 85], dtype=float)

# Standardize features
mean = X.mean(axis=0)
std = X.std(axis=0)
X_scaled = (X - mean) / std

# Initialize parameters
n_samples, n_features = X_scaled.shape
w = np.zeros(n_features)
b = 0.0

learning_rate = 0.01
epochs = 10000

# Gradient Descent
for epoch in range(epochs):
    y_pred = X_scaled @ w + b
    error = y_pred - y

    dw = (2 / n_samples) * (X_scaled.T @ error)
    db = (2 / n_samples) * np.sum(error)

    w -= learning_rate * dw
    b -= learning_rate * db

# Evaluate
y_pred = X_scaled @ w + b
mse = np.mean((y - y_pred) ** 2)

print("Weights:", w)
print("Bias:", b)
print("Predictions:", np.round(y_pred, 2))
print("MSE:", round(mse, 4))
```

```python
new_student = np.array([[4, 85, 75]], dtype=float)

new_student_scaled = (new_student - mean) / std

prediction = new_student_scaled @ w + b

print("Predicted marks:", prediction[0])
```
